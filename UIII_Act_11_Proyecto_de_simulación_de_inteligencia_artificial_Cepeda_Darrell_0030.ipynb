## Ejemplos Carpintero

```python
# --- SISTEMA DE IA: GESTOR INTELIGENTE DE CARPINTERÍA ---

# Se inicia con una lista vacía para almacenar los registros
listado_carpinteros = []

def mostrar_menu():
    """Muestra las opciones del sistema interactivo"""
    print("\n" + "="*50)
    print("      CENTRO DE MANDO IA: RED DE CARPINTEROS")
    print("="*50)
    print("1. Registrar nuevo Carpintero (Entrada de datos)")
    print("2. Ver listado de personal (Tabla organizada)")
    print("3. Ejecutar diagnóstico IA de capacidad")
    print("4. Salir")

def registrar():
    """Crea un diccionario vacío y lo llena con la entrada del usuario"""
    # El nombre del diccionario empieza vacío en cada registro
    carpintero = {}

    print("\n--- REGISTRO DE NUEVO PERFIL ---")
    # Entrada interactiva de datos (Key: Values)
    nombre = input("Nombre del carpintero: ")
    madera = input("Tipo de madera preferida: ")
    maquinaria = input("¿Cuenta con maquinaria propia? (si/no): ").lower()
    muebles = input("Catálogo de muebles (ej. Sillas, Mesas, Camas): ")
    taller = input("Ubicación o nombre del taller: ")
    pedidos = int(input("Cantidad de pedidos pendientes: "))
    presupuesto = float(input("Costo de presupuesto promedio ($): "))

    # Asignación de pares clave-valor al diccionario
    carpintero["nombre"] = nombre
    carpintero["madera"] = madera
    carpintero["maquinaria"] = maquinaria
    carpintero["catalogo"] = muebles
    carpintero["taller"] = taller
    carpintero["pedidos"] = pedidos
    carpintero["costo"] = presupuesto

    # Se agrega el diccionario a la lista global
    listado_carpinteros.append(carpintero)
    print(f"\nIA: Perfil de '{nombre}' integrado con éxito.")

def imprimir_tabla():
    """Presenta el listado en forma de tabla bien alineada usando f-strings"""
    if not listado_carpinteros:
        print("\nIA ALERTA: La base de datos está vacía.")
    else:
        print("\n" + "-"*100)
        # Encabezado con buena distribución de información
        header = f"{'Nombre':<15} | {'Madera':<12} | {'Máq.':<5} | {'Pedidos':<8} | {'Presupuesto':<10} | {'Taller':<15}"
        print(header)
        print("-" * 100)

        # Ciclo for para recorrer la lista de diccionarios
        for c in listado_carpinteros:
            print(f"{c['nombre']:<15} | {c['madera']:<12} | {c['maquinaria']:^5} | {c['pedidos']:^8} | ${c['costo']:<10.2f} | {c['taller']:<15}")

def ejecutar_ia():
    """Simula lógica de IA mediante sistemas de reglas condicionales"""
    print("\n--- ANALIZANDO ESTADO DE LA RED MEDIANTE IA ---")
    if not listado_carpinteros:
        print("IA ERROR: No hay datos para procesar.")
    else:
        for c in listado_carpinteros:
            print(f"\n> Análisis para: {c['nombre']} ({c['taller']})")

            # Lógica de toma de decisiones (Inteligencia Artificial Básica)
            if c["pedidos"] > 5 and c["maquinaria"] == "no":
                print("IA DECISIÓN: RIESGO DE INCUMPLIMIENTO.")
                print("Sugerencia: El carpintero requiere inversión en maquinaria para desahogar pedidos.")

            elif c["pedidos"] > 10 and c["maquinaria"] == "si":
                print("IA DECISIÓN: SATURACIÓN DE TRABAJO.")
                print("Sugerencia: Derivar nuevos clientes a otro taller de la red.")

            elif c["costo"] > 5000 and c["madera"] == "caoba":
                print("IA DECISIÓN: PERFIL DE EBANISTERÍA PREMIUM.")
                print("Sugerencia: Asignar exclusivamente proyectos de alta gama.")

            elif c["pedidos"] <= 2:
                print("IA DECISIÓN: TALLER CON DISPONIBILIDAD INMEDIATA.")
                print(f"Acción: Promocionar su catálogo de {c['catalogo']} para aumentar ventas.")

            else:
                print("IA DECISIÓN: OPERACIÓN NORMAL.")
                print("Estado: Parámetros de eficiencia dentro del rango esperado.")

# --- CICLO PRINCIPAL DE INTERACCIÓN ---
# Uso de while True para mantener el menú activo
while True:
    mostrar_menu()
    opcion = input("\nSeleccione una opción (1-4): ")

    if opcion == "1":
        registrar()
    elif opcion == "2":
        imprimir_tabla()
    elif opcion == "3":
        ejecutar_ia()
    elif opcion == "4":
        print("\nCerrando sistema de IA... ¡Seguridad y precisión en el taller!")
        break # Sale del bucle infinito
    else:
        print("Opción no válida. Intente de nuevo.")

print("------------------------------------------")
print("Simulación de proceso lógico finalizada.")

print("Programa realizado por Cepeda Darrell NL = 10")
```
